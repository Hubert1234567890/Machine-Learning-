#Feature Engineering Codes

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#  1. feature scaling

i. standardiation

```python
df = pd.read_csv('/content/Social_Network_Ads.csv')
df
```

```python
df.iloc[:,2:].sample(5)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test= train_test_split(df.iloc[:,2:4],df.iloc[:,4],test_size =0.3,random_state = 42)
print(x_train.shape)
print(x_test.shape)
```

now we do the standardization

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(x_train)
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled_df = pd.DataFrame(x_train_scaled, columns=x_train.columns)
display(x_train_scaled_df.head())


x_test_scaled_df = pd.DataFrame(x_test_scaled, columns=x_test.columns)
display(x_test_scaled_df.head())
```

```python
np.round(x_train_scaled_df.describe(),1)
```

```python
np.round(x_train.describe(),1)
```

scatter plot of the values before and after standardization

```python
plt.figure(figsize =(10,4))

# First plot (Left)
plt.subplot(1,2,1)
plt.scatter(x_train['Age'],x_train['EstimatedSalary'])
plt.title('Before Scaling')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')

# Second plot (Right)
plt.subplot(1,2,2)
plt.scatter(x_train_scaled_df['Age'],x_train_scaled_df['EstimatedSalary'],color='red')
plt.title('After Scaling')
plt.xlabel('Age (scaled)')
plt.ylabel('Estimated Salary (scaled)')

# Display both together
plt.show()
```

```python
plt.figure(figsize = (10,4))

plt.subplot(1,2,1)
sns.kdeplot(x_train['Age'])
sns.kdeplot(x_train['EstimatedSalary'])
plt.title('Before scaling')


plt.subplot(1,2,2)
sns.kdeplot(x_train_scaled_df['Age'])
sns.kdeplot(x_train_scaled_df['EstimatedSalary'])
plt.title('Before scaling')
```

```python
plt.figure(figsize = (10,4))

plt.subplot(1,2,1)
sns.kdeplot(x_train['Age'])
plt.title('Distribution before scaling Age')


plt.subplot(1,2,2)
sns.kdeplot(x_train_scaled_df['Age'])
plt.title('Distribution after scaling Age')
```

```python
plt.figure(figsize = (10,4))

plt.subplot(1,2,1)
sns.kdeplot(x_train['EstimatedSalary'])
plt.title('Distribution before scaling Estimatedsalary')


plt.subplot(1,2,2)
sns.kdeplot(x_train_scaled_df['EstimatedSalary'])
plt.title('Distribution after scaling EstimatedSalary')
```
